# BFS

```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        traversal.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

    return traversal

result = bfs(graph, 'A')
print("BFS Traversal:")
print(" -> ".join(result))
```

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        traversal.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

    return traversal

result = bfs(graph, 'A')
print("BFS Traversal:")
print(" -> ".join(result))

G = nx.DiGraph()

for node in graph:
    for neighbour in graph[node]:
        G.add_edge(node,neighbour)

pos={
        'A': (0, 3),
    'B': (-0.5, 2),
    'C': (0.5, 2),
    'D': (-0.9, 1),
    'E': (-0.23, 1),
    'F': (0, 1),
    'G': (-0.5, 0)
}
plt.figure(figsize=(4,3))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=800,
    node_color='green',
    font_size=10,
    font_weight='bold',
    arrows=True,
    arrowsize=13,
    edge_color='black',
    width=2,
)

plt.title("BFS")
plt.axis('off')
plt.show()
```
